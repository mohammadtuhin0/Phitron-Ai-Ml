# 1. Diabetes Dataset – Missing Value Imputation & Encoding

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv("diabetes.csv")

df
```

```python
df.isnull().sum()
```

```python
invalid_cols = ['Glucose', 'BloodPressure', 'BMI']

df[invalid_cols] = df[invalid_cols].replace(0, np.nan)
```

```python
df.isnull().sum()
```

```python
df['Glucose'] = df['Glucose'].fillna(df['Glucose'].median())
df['BloodPressure'] = df['BloodPressure'].fillna(df['BloodPressure'].median())
df['BMI'] = df['BMI'].fillna(df['BMI'].median())
```

```python
df
```

```python
df.isnull().sum()
```

```python
def age_group(age):
    if age < 30:
        return "Young"
    elif age <= 50:
        return "Adult"
    else:
        return "Senior"

df['Age_Group'] = df['Age'].apply(age_group)
```

```python
df
```

```python
mapping = {
    'Young': 0,
    'Adult': 1,
    'Senior': 2
}

df['Age_Group_Encoder'] = df['Age_Group'].map(mapping)
```

```python
df
```

# 2. Heart Disease Dataset – One Hot Encoding

```python
df = pd.read_csv("heart.csv")

df
```

```python
df.info()
```

```python
print("Before Encoding:", df.shape)
```

```python
cp_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

cp_ohe.fit(df[['cp']])
```

```python
encoder_df = cp_ohe.transform(df[['cp']])
df = pd.concat([df, encoder_df], axis=1)
```

```python
df
```

```python
thal_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')
thal_ohe.fit(df[['thal']])

thal_encoder = thal_ohe.transform(df[['thal']])
df = pd.concat([df, thal_encoder], axis=1)
```
